```python
import wbdata
import pandas as pd
import plotly.express as px

# المؤشرات المهمة (نفس اللي قبل كده)
indicators = {
    "SE.PRM.ENRR": "معدل الالتحاق بالتعليم الابتدائي (% gross)",
    "SE.SEC.ENRR": "معدل الالتحاق بالتعليم الثانوي (% gross)",
    "SE.TER.ENRR": "معدل الالتحاق بالتعليم العالي (% gross)",
    "SE.ADT.LITR.ZS": "معدل محو الأمية للبالغين (% من عمر 15+)"
}

# تحميل البيانات لمصر (كود البلد EGY) من 2000 لحد أحدث سنة متاحة
df = wbdata.get_dataframe(indicators, country="EGY")

# تنظيف البيانات (إزالة الصفوف الفاضية وترتيب)
df = df.reset_index()
df = df.dropna(how='all', subset=list(indicators.values()))
df = df.sort_values(by='date', ascending=True)

# عرض الجدول (آخر 15 سنة عشان يبقى واضح)
print("بيانات مؤشرات التعليم في مصر (أحدث السنوات المتاحة):")
display(df.tail(15))
```

```python
# عنوان الداشبورد
print("داشبورد تفاعلي لمؤشرات التعليم في مصر نحو SDG4")

# رسم بياني 1: خطوط زمنية لكل المؤشرات
fig1 = px.line(df, x='date', y=list(indicators.values()),
               title='تطور مؤشرات التعليم في مصر (2000 - أحدث سنة)',
               labels={'value': 'النسبة المئوية %', 'date': 'السنة', 'variable': 'المؤشر'})
fig1.update_layout(hovermode='x unified', template='plotly_white')
fig1.show()

# رسم بياني 2: أعمدة لأحدث سنة متاحة
latest_year = df['date'].max()
latest_data = df[df['date'] == latest_year].melt(id_vars='date', value_vars=list(indicators.values()))
fig2 = px.bar(latest_data, x='variable', y='value',
              title=f'مؤشرات التعليم في أحدث سنة ({latest_year})',
              labels={'value': 'النسبة %', 'variable': 'المؤشر'})
fig2.update_layout(template='plotly_white')
fig2.show()
```
